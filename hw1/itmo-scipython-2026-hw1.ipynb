# How to encode image to txt file and encode txt file to image?

Specify list of files

```python
os.listdir()
```

## Step 1. Use function `img_to_txt` to encode img to txt

```python
lst_img = ['itmo-scipython-2026-hw1-img1.jpg', 'itmo-scipython-2026-hw1-img2.jpg']

def img_to_txt(lst_img):
    for img in lst_img:
        with open(img, mode='rb') as img_file:
            encoded_img = base64.b64encode(img_file.read()).decode(encoding='utf-8')
        text_path = img[:-4] + '.txt'
        with open(text_path, mode='w') as txt_file:
            txt_file.write(encoded_img)
    return None

img_to_txt(lst_img)
```

## Step 2. Use function `txt_to_img` to decode txt

```python
lst_txt = ['itmo-scipython-2026-hw1-img1.txt', 'itmo-scipython-2026-hw1-img2.txt']

def txt_to_img(lst_txt):
    for txt in lst_txt:
        with open(txt, mode='r') as txt_file:
            decoded_img = base64.b64decode(txt_file.read())
        path_to_restored_img = txt[24:28]+'_restored.jpg'
        with open(path_to_restored_img, mode='wb') as new_img:
            new_img.write(decoded_img)
    return None

txt_to_img(lst_txt)
```
